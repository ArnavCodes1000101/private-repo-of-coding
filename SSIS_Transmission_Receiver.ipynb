```python
def word_to_binary(word):
    binary_representation = ' '.join(format(ord(char), '08b') for char in word)
    return binary_representation

# Example usage:
input_word = "arnav"
binary_output = word_to_binary(input_word)
print(f"The word '{input_word}' in binary is: {binary_output}")
```
